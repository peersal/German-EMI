```python
from sentence_transformers import SentenceTransformer
from sentence_transformers import util
from sentence_transformers import models
from sentence_transformers.models import WordEmbeddings
from sentence_transformers.models.tokenizer import WhitespaceTokenizer
from nltk.corpus import stopwords
```

```python
def sentence_transformers_model(embeddings_file, output_path, stopwords=[]):
    embedding_model = WordEmbeddings.from_text_file(embeddings_file, tokenizer=WhitespaceTokenizer(stop_words=stopwords))
    pooling_model = models.Pooling(embedding_model.get_word_embedding_dimension())#mean pooling                                                         
    model = SentenceTransformer(modules=[embedding_model, pooling_model])
    model.save(output_path)
```

```python
german_stopwords = list(stopwords.words('german'))

w2v_embedding_path = "/kaggle/input/w2v-txt/word2vec_new.txt"
output_path = "/kaggle/working/model"

sentence_transformers_model(w2v_embedding_path,output_path, stopwords=german_stopwords)
```
